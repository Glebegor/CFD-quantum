# Rozdělení kvantových registrů pro QLBM D2Q9

Výpočet používá prostorovou mřížku $512 \times 512$ a rychlostní model D2Q9. Kvantový stav proto musí obsahovat dva souřadnicové registry a jeden registr směru:

$$|x\rangle|y\rangle|d\rangle.$$

## Souřadnicový registr $x$

Registr musí rozlišit všech 512 pozic na ose $x$. Pro registr s $n$ qubity existuje $2^n$ bázových stavů. Platí:

$$2^8 = 256 < 512, \qquad 2^9 = 512.$$

Proto registr $|x\rangle$ potřebuje **9 qubitů** a reprezentuje indexy buněk $x \in \{0,\ldots,511\}$.

## Souřadnicový registr $y$

Osa $y$ má také 512 pozic, takže stejným způsobem:

$$n_y = \lceil\log_2(512)\rceil = 9.$$

Registr $|y\rangle$ tedy rovněž potřebuje **9 qubitů** a reprezentuje $y \in \{0,\ldots,511\}$.

## Registr směru $d$

Model D2Q9 obsahuje devět diskrétních rychlostí: jeden klidový stav, čtyři osové směry a čtyři diagonální směry. Tři qubity nestačí, protože $2^3=8<9$. Čtyři qubity poskytují $2^4=16$ stavů, a proto:

$$n_d = \lceil\log_2(9)\rceil = 4.$$

| Stav $d$ | Binárně | Směr | Vektor $(c_x,c_y)$ |
|---:|:---:|---|:---:|
| 0 | `0000` | klid | $(0,0)$ |
| 1 | `0001` | východ | $(1,0)$ |
| 2 | `0010` | sever | $(0,1)$ |
| 3 | `0011` | západ | $(-1,0)$ |
| 4 | `0100` | jih | $(0,-1)$ |
| 5 | `0101` | severovýchod | $(1,1)$ |
| 6 | `0110` | severozápad | $(-1,1)$ |
| 7 | `0111` | jihozápad | $(-1,-1)$ |
| 8 | `1000` | jihovýchod | $(1,-1)$ |

Stavy 9 až 15 nejsou v D2Q9 fyzikálně využité. Kvantový obvod je nesmí zamíchat s platnými směrovými stavy.

## Celkový počet qubitů

$$n_{\mathrm{celkem}} = n_x+n_y+n_d = 9+9+4 = \boxed{22\ \mathrm{qubitů}}.$$

Těchto 22 qubitů popisuje pouze polohu a směr. Pomocné qubity potřebné pro překážku, kolizi, řízení operací nebo aritmetiku se počítají zvlášť.

```python
import math

nx = 512
ny = 512
number_of_directions = 9  # D2Q9

qubits_x = math.ceil(math.log2(nx))
qubits_y = math.ceil(math.log2(ny))
qubits_direction = math.ceil(math.log2(number_of_directions))
total_qubits = qubits_x + qubits_y + qubits_direction

print(f"Registr x:      {qubits_x} qubitů ({2**qubits_x} stavů)")
print(f"Registr y:      {qubits_y} qubitů ({2**qubits_y} stavů)")
print(f"Registr směru:  {qubits_direction} qubity ({2**qubits_direction} stavů, z toho 9 použitých)")
print(f"Celkem:         {total_qubits} qubitů")
```
